Usando una clase podemos generalizar el objeto anterior (llamemosla spring). Los parametros que esta deberá
tener al inicializarse deberán ser el objeto ax de matplotlib (para indicarle en que plot estará la ilustración);
la posición y altura de los soportes cuando el resorte está en equilibrio (por default serán (0.5,0.1) y (0.5,0.8), como el la figura anterior); un valor de escala (por default 1, e indicara el grosor del resorte, largo de los soportes, y junto con el parametro anterior, determinara cuantas líneas diagonales tiene el resorte); un valor de peso para el grosor de las líneas (1 por default), y un valor de compresión de 0 a 1 (0.5 default), donde 0 el resorte comprimido a la mitad (mínimo disponible), 1 es el expandido a 1.5 (máximo disponible). En las próximas celdas se irá construyendo poco a poco la clase y unas pruebas para determinar su funcionamiento.

```python
#plano
fig, ax = plt.subplots()
fig.set_size_inches(4,6)
ax.set_xlim(0,1)
ax.set_ylim(0,1.5)

class Resorte:
    #función init
    def __init__(self, ax, pos=[(0.5,0.1),(0.5,0.8)], scale=1, weight=1, compression=0.5):
        self.ax = ax
        self.pos = pos
        self.scale = scale
        self.weight = weight
        self.compression = compression

    #función para dibujar el resorte (Primero con los valores en default)
    def draw(self):
        pos = self.pos
        l=self.scale*0.04
        # por default, cada linea diagonal, sin comprimir, y a escala 1 será de de 0.2x0.1, y el soporte será de
        # 0.1 de largo, de tal manera que si n es la cantidad de diagonales, n+2 será el largo del resorte.
        # (restringiendo la longitud al conjunto (n+2)0.1 con n natural, adelante se modificará el código
        # para que se permitan valores continuos de longitud)
        # Para este caso, determinamos el número de líneas diagonales + 2 (para contemplar el espacio de
        # los soportes )
        n = int((self.pos[1][1] - self.pos[0][1]+0.001)/0.1 )
        #con el número de líneas diagonales, n, las guardamos en un array
        self.diag = []
        for i in range(1,n-1):
            self.diag.append( ax.plot([pos[0][0]-0.1,pos[1][0]+0.1],[i*0.1+pos[0][1], (i+1)*0.1+pos[0][1]],
                                 color="black", lw=3)[0] )
        # con ese mismo numero podemos dibujar las líneas horizontales
        self.horz = []
        for i in range(1,n):
            self.horz.append( ax.plot( [pos[0][0]-l/2,pos[1][0]+l/2], [i*0.1+pos[0][1], i*0.1+pos[0][1]],
                                 color = "black", lw=1 ) )
        #finalmente dibujamos los soportes
        self.supph1, = ax.plot( [pos[0][0]-l,pos[1][0]+l], [pos[0][1], pos[0][1]], color="black", lw=3)
        self.suppv1, = ax.plot( [pos[0][0], pos[0][0]], [pos[0][1], pos[0][1]+0.1],color="black", lw=3)
        self.supph2, = ax.plot( [pos[0][0]-l,pos[1][0]+l], [pos[1][1], pos[1][1]], color="black", lw=3)
        self.suppv2, = ax.plot( [pos[1][0], pos[1][0]], [pos[1][1], pos[1][1]-0.1],color="black", lw=3)



#probamos dibujar con 7 lineas diagonales empezando en otra posicion
resorte = Resorte(ax, pos=[(0.2,0.0),(0.2, 0.9)])
resorte.draw()



```

Ahora conviene hacer m'as din'amico el co'digo, de tal manera que no estemos restringidos a poner una posicio'n que
limite la longitud al conjunto $\{ (n+2)0.1, n \in N\}$. Para eso podemos hacer un poco mas flexible el codigo de tal manera que,
si por ejemplo, pusieramos una posicion que de una altura entre 0.7 y 0.8, si esta es mayor a .75, que se reondee a 6 (8-2) lineas
diagonales, y si es menor a 0.75, que se redondee hacia abajo, a 5 (7-2) lineas diagonales. Entonces:

```python
#plano
fig, ax = plt.subplots()
fig.set_size_inches(4,6)
ax.set_xlim(0,1)
ax.set_ylim(0,1.5)

class Resorte:
    #función init
    def __init__(self, ax, pos=[(0.5,0.1),(0.5,0.8)], scale=1, weight=1, compression=0.5):
        self.ax = ax
        self.pos = pos
        self.scale = scale
        self.weight = weight
        self.compression = compression

    #función para dibujar el resorte (Primero con los valores en default)
    def draw(self):
        pos = self.pos
        l=self.scale*0.04
        # por default, cada linea diagonal, sin comprimir, y a escala 1 será de de 0.2xA,
        # donde A sera igual a la altura de la linea diagonal
        # de tal manera que si n es la cantidad de diagonales, h = nA+0.2 será el largo del resorte.
        # si la altura del resorte es h, h-0.2 sera la de las lineas diagonales
        # Para determinar n, primero dividimos 0.1 (alto ideal de la diagonal) entre h-0.2 y redondeamos hacia abajo.
        # si el residuo es mayor a 0.05 (media altura ideal), a n le agregamos uno. Al final el alto real de las diagonales
        # sera (h-0.2)/n

        #determianos la altura base del resorte para este caso simple
        h = self.pos[1][1] - self.pos[0][1] +0.00001

        #determinamos n con esa condicion
        n = int((h-0.2)/0.1) if (h-0.2)%0.1<0.05 else int((h-0.2)/0.1)+1
        #definimos la altura real de las diagonales
        hd = (h-0.2)/n

        print(h,n,hd)

        #con el número de líneas diagonales, n, las guardamos en un array
        self.diag = []
        #notese que en la clase anterior n contaba a los soportes. ahora no
        for i in range(1,n+1):
            self.diag.append( ax.plot([pos[0][0]-0.1,pos[1][0]+0.1],[i*hd+pos[0][1], (i+1)*hd+pos[0][1]],
                                 color="black", lw=3)[0] )
        # con ese mismo numero podemos dibujar las líneas horizontales
        self.horz = []
        for i in range(1,n+2):
            self.horz.append( ax.plot( [pos[0][0]-l/2,pos[1][0]+l/2], [i*hd+pos[0][1], i*hd+pos[0][1]],
                                 color = "black", lw=1 ) )
        #finalmente dibujamos los soportes
        self.supph1, = ax.plot( [pos[0][0]-l,pos[1][0]+l], [pos[0][1], pos[0][1]], color="black", lw=3)
        self.suppv1, = ax.plot( [pos[0][0], pos[0][0]], [pos[0][1], pos[0][1]+0.1],color="black", lw=3)
        self.supph2, = ax.plot( [pos[0][0]-l,pos[1][0]+l], [pos[1][1], pos[1][1]], color="black", lw=3)
        self.suppv2, = ax.plot( [pos[1][0], pos[1][0]], [pos[1][1], pos[1][1]-0.1],color="black", lw=3)



#probamos dibujar con 7 lineas diagonales empezando en otra posicion
resorte = Resorte(ax, pos=[(0.2,0.0),(0.2, 0.96)])
resorte.draw()

```

Finalmente faltaria ajustar la compresion del resorte, pero veamos que esto es como aplicar una escala nuevamente, pero solo
en la altura del objeto. Para ello, primero definamos una variable $s = 0.5 +$ compression, de tal manera que $s$ va de 0.5 a 1.5.

El ultimo problema a resolver es el de la posición del resorte, pues la hemos limitado solo a una en la que
esté orientado verticalmente, y al probar el código de otra forma, el dibujo no se ve como debería. Para resolver esto vease un detalle importante: el proceso de ilustración depende principalmente de la altura del resorte, y esta la podemos obtener solo tomando la distancia euclidieana entre las dos posiciones del parametro _pos_ . Redefiniendo la altura del resorte de esta forma, el código aun solo nos permite hacer el resorte de forma vertical, ¿Cómo cambiamos esto? podemos conservar el mismo código, pero corrigiendo las posiciones a través de dos transformaciones:
una traslación, y una rotación.

Trabajemos primero con la transformación de rotación, dibujando el soporte fijo del resorte en el origen.

```python
#plano
fig, ax = plt.subplots()
fig.set_size_inches(8,6)
ax.set_xlim(-1,1)
ax.set_ylim(0,1.5)

class Resorte:
    #función init
    def __init__(self, ax, pos=[(0.5,0.1),(0.5,0.8)], scale=1, weight=1, compression=0.5):
        self.ax = ax
        self.pos = pos
        self.scale = scale
        self.weight = weight
        self.compression = compression

    #función para dibujar el resorte (Primero con los valores en default)
    def draw(self):
        pos = self.pos
        # por default, cada linea diagonal, a escala E será de de 0.2*E x A,
        # donde A sera igual a la altura de la linea diagonal
        # de tal manera que si n es la cantidad de diagonales, h = nA+0.2*E será el largo del resorte.
        # si la altura del resorte es h, h-0.2*E sera la de las lineas diagonales
        # Para determinar n, primero dividimos ih = 0.1*E (alto ideal de la diagonal) entre h-2*ih y redondeamos hacia abajo.
        # si el residuo es mayor a ih/2 (media altura ideal), a n le agregamos uno. Al final el alto real de las diagonales
        # sera (h-2*ih)/n

        # definimos una variable para ajustar la compresion
        s = (0.5+self.compression)
        # definimos la altura del resorte con la norma euclidea
        h = np.sqrt( (pos[0][0] - pos[1][0])**2 + (pos[0][1] - pos[1][1])**2 )*s +0.00001

        
        #determinamos grueso y altura diagonal ideal
        
        ih = 0.1*self.scale*s
        l = 0.4*ih/s

        #determinamos n con esa condicion
        n = int((h-2*ih)/ih) if (h-2*ih)%ih<ih/2 else int((h-2*ih)/ih)+1
        #definimos la altura real de las diagonales
        hd = (h-2*ih)/n

        # definimos una función para hacer la rotación de un vecor
        def R(x,y,pos = pos):
            #definimos el angulo de la rotación
            theta = np.arctan( (pos[1][1]-pos[0][1])/(pos[1][0]-pos[0][0]) ) + np.pi/2
            theta = theta if (pos[1][1]-pos[0][1])>0 else theta-np.pi
            return np.matmul([x,y],[[np.cos(theta), np.sin(theta)],[-np.sin(theta),np.cos(theta)]])


        #con el número de líneas diagonales, n, las guardamos en un array
        self.diag = []
        #dibujamos las líneas diagonales, pero dibujandolo desde R(0,0) hasta R(0,h) con R la transformación
        # de rotación
        for i in range(1,n+1):
            self.diag.append( ax.plot([R(-ih/s,i*hd)[0],R(ih/s,(i+1)*hd)[0]],[R(-ih/s,i*hd)[1],R(ih/s,(i+1)*hd)[1]],
                                 color="black", lw=3*self.weight)[0] )
        # con ese mismo numero podemos dibujar las líneas horizontales
        self.horz = []
        for i in range(1,n+2):
            self.horz.append( ax.plot( [R(-l/2,i*hd)[0],R(l/2,i*hd)[0]], [R(-l/2,i*hd)[1],R(l/2,i*hd)[1]],
                                 color = "black", lw=1*self.weight ) )

        #finalmente dibujamos los soportes
        self.supph1, = ax.plot( [R(-1,0)[0], R(l,0)[0]], [R(-1,0)[1], R(l,0)[1]], color="black", lw=3*self.weight)
        self.suppv1, = ax.plot( [0,R(0,ih)[0]], [0,R(0,ih)[1]],color="black", lw=3*self.weight)
        self.supph2, = ax.plot( [R(-l,h)[0], R(l,h)[0]], [R(-l,h)[1], R(l,h)[1]], color="black", lw=3*self.weight)
        self.suppv2, = ax.plot( [R(0,h)[0], R(0,h-ih)[0]], [R(0,h)[1], R(0,h-ih)[1]],color="black", lw=3*self.weight)



#definimos dos posiciones y dibujamos la línea donde debería ir el resorte, y el resorte solo con la rotación
# X del soporte fijo, y X del soporte movil
posx = [0.6, 0.2]
posy = [0.8, 1.3]
plt.plot(posx, posy)
resorte = Resorte(ax, pos=[(posx[0],posy[0]),(posx[1], posy[1])], scale=1, weight=1, compression=0.5)
resorte.draw()

```
